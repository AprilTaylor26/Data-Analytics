```python
#Imports required
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

#Read dataset
df = pd.read_excel("C:\\Files\\Excel\\StudentsPerformance.xlsx")
df.head()
```

```python
# Historgram of distributed math scores

plt.figure(figsize=(8,5))
plt.hist(df["math score"], bins=15, edgecolor="black")
plt.title("Distribution of Math Scores")
plt.xlabel("Math Score")
plt.ylabel("Frequency")
plt.show()
```

```python
# Bar chart of Average math score by race/ethnicty 

avg_by_race = df.groupby("race/ethnicity")["math score"].mean().sort_values()

plt.figure(figsize=(8,5))
sns.barplot(x=avg_by_race.index, y=avg_by_race.values, palette="viridis")
plt.title("Average Math Score by Race/Ethnicity")
plt.xlabel("Race/Ethnicity Group")
plt.ylabel("Average Math Score")
plt.show()
```

```python
#Scatter plot of math vs writing score

plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x="math score", y="writing score")
plt.title("Math Score vs. Writing Score")
plt.xlabel("Math Score")
plt.ylabel("Writing Score")
plt.show()

```

```python
#Heatmap of correlations

num_cols = df[["math score", "reading score", "writing score"]]

plt.figure(figsize=(6,4))
sns.heatmap(num_cols.corr(), annot=True, cmap="coolwarm", linewidths=0.5)
plt.title("Correlation Heatmap of Student Scores")
plt.show()

```
